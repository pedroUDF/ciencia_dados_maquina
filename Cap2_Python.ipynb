```python
cidades["População"] = cidades["População"].astype('float64')
cidades.dtypes
```

```python
cidades["Data"] = ["3-8-2021", "3-9-2021", "3-10-2021", "3-11-2021", "3-12-2021"]
# Setando a data para a forma que conhecemos (acho que nao funcionou nao kkk)
cidades["Data"] = pd.to_datetime(cidades["Data"], format='%d-%m-%Y')

cidades
```

```python
# Agora sim, seta a data no formato que conhecemos.
cidades["Data"] = cidades["Data"].dt.strftime('%d-%m-%Y') # Dia, Mês, Ano

cidades
```

```python
# Renomeando Colunas
cidades.rename(columns= {"População": "N° Habitantes", "Data": "Data - Censo"}, inplace=True) #inlpace=true serve para mudar na própria tabela, por padrão o rename cria uma cópia da tabela

cidades
```

```python
# Selecionando colunas com os nomes especificados
cidadesHab = cidades[["N° Habitantes", "Salário-Médio"]]

cidadesHab
```

```python
#selecionando colunas, no exemplo, selecionei a 1 e a 3
cidades.loc[[1,3]]
```

```python
# Fatiamento do Py - o último num não entra no fatiamento.
cidades.iloc[1:5]
```

```python
#  Seleciona todas as linhas e as duas primeiras colunas ( o ":" significa 'pegue tudo, selecione todas as linhas')
cidades.iloc[:, 0:2]
```

```python
# :3 == 0:3
cidades.iloc[:3, :3]
```

```python
# Seleciona todas as linhas e as colunas determinadas
cidades.iloc[:, [1, 3, 0, 2]]
```

```python
# Estrutura do insert = df.insert(posição, "nome da coluna", ["valores"])
cidades.insert(2, "Sigla", ["PR", "SP", "SC", "PR", "SP"]) # Adiciona novos dados no índice dois
cidades.insert(2, "Sigla-Repetida", cidades["Sigla"]) # Adiciona uma coluna repetida "Sigla"

cidades
```

```python
# Excluir coluna duplicada
cidades = cidades.drop(['Sigla-Repetida'], axis=1)
# ou cidades = cidades.pop("Sigla-Repetida")

cidades
```

```python
# Consulta - df.query("nome coluna/linha + condição")
# As aspas são para evitar possíveis confusões, evitam o Py fazer a operação: Salário - Médio
cidades.query("`Salário-Médio` > 1400.0")
```

```python
# Pode se usar o "and" ou "or" para consultar mais de uma coisa ao mesmo tempo
cidades.query("`N° Habitantes` > 500000 and `N° Habitantes` < 1300000")
```

```python
# Ordenação (sempre em ordem crescente)
cidades.sort_values(by="Cidade", inplace=True)

cidades
```

```python
cidades.sort_values(by="Salário-Médio", inplace=True)

cidades
```

```python
# Juntando Dataframes, mas agora, criando um
cidades_salarios = pd.DataFrame(
    [
        ["Paraná", "Londrina", 1356.00],
        ["São Paulo", "São Carlos", 1508.00],
        ["Santa Catarina", "Florianópolis", 1798.00],
        ["Paraná", "Curitiba", 2293.00],
        ["São Paulo", "Campinas", 1710.00]
    ], columns=["Estado", "Cidade", "Salário-Médio"])
cidades_salarios
```

```python
# Criando mais uma tabela cidades_pop
cidades_pop = pd.DataFrame(
    [
        ["Londrina", 575377],
        ["São Carlos", 254484],
        ["Florianópolis", 508826],
        ["Curitiba", 1963726],
        ["Campinas", 1223237]
    ], columns=['Cidade', 'Habitantes'])
cidades_pop
```

```python
# Mesclando as tabelas
# on="Cidade" = Especifica qual coluna serve como chave de ligação
pd.merge(left=cidades_salarios, right=cidades_pop, on="Cidade")

#Por padrão, o pd.merge só traz as linhas onde a cidade existe em ambas as tabelas
```

```python
# how="outer" = Nenhuma linha é omitida, no anterior, tem um comando 'escondido' = how="inner"
pd.merge(left=cidades_salarios, right=cidades_pop, on="Cidade", how="outer")
```

```python
# Concatenação, parece com o anterior
cidades_df1 = pd.DataFrame(
    [
        ["Londrina", 575377],
        ["São Carlos", 254484],
        ["Florianópolis", 508826],
    ], columns=["Cidade", "Habitantes"])

cidades_df1
```

```python
cidades_df2 = pd.DataFrame(
    [
        ["Curitiba", 1963726],
        ["Campinas", 1223237]
    ], columns=["Cidade", "Habitantes"])

cidades_df2
```

```python
pd.concat([cidades_df1, cidades_df2], ignore_index=True)
# ignore_index =True: Refaz a contagem numérica das linhas (os índices) do zero (0, 1, 2, 3, 4).
```
